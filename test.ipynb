```python
import http.client
import json
import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine , text


load_dotenv()  

api_key = os.getenv("GAS_API")

if not api_key:
    raise ValueError("GAS_API not found in .env")

conn = http.client.HTTPSConnection("api.collectapi.com")

headers = {
    'content-type': "application/json",
    'authorization': f"apikey {api_key}"
    }

conn.request("GET", "/gasPrice/stateUsaPrice?state=WA", headers=headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))
```

```python
parsed_data = json.loads(data.decode("utf-8"))
parsed_data
```

```python
type(parsed_data)
```

```python
cities = parsed_data['result']['cities']
type(cities)
```

```python
cities
```

```python
cities_df = pd.DataFrame(cities)
cities_df
```

```python
cities_df.drop(columns=['lowername'],inplace=True)
cities_df
```

```python
cities_df.rename(columns={'name':'cities'}, inplace=True)
cities_df
```

```python
DB_HOST = os.getenv('DB_HOST')
DB_NAME = os.getenv('DB_NAME')
DB_PORT = os.getenv('DB_PORT')
DB_USER = os.getenv('DB_USER')
DB_PASSWORD = os.getenv('DB_PASSWORD')

engine = create_engine(f'postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}')
```

```python
cities_df.to_csv('Data/gas_prices.csv', index=False)
```

```python
# checking if our connection is ok
with engine.connect() as conn:
    result = conn.execute(text('select 1;'))
    for i in result:
        print(i)
```

```python
cities_df.to_sql('gas_prices', engine , if_exists='replace', index=False)
```
